## Checking Multiple Model with Hyper Parameter Tunning and their Accuracy

```python
from sklearn import svm
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_params = {
    'svm' : {
        'model' : svm.SVC(gamma= 'auto'),
        'params' : {
            'C' : [0.1, 1, 10],
            'kernel' : ['rbf', 'linear']
            
        }
    },
    'random_forest' : {
        'model' : RandomForestClassifier(),
        'params' : {
            'n_estimators' :[50, 100, 200],
            'max_depth': [None, 10, 20]
        }
    },
    'logistic_regression' : {
        'model' : LogisticRegression(solver = 'liblinear', multi_class = 'auto'),
        'params' : {
            'C' : [0.1, 1, 10]
        }
    },
    'Decision Tree' : {
        'model' : DTC(random_state=42),
         'params' : {
            'max_depth': [None, 10, 20],
             'min_samples_split': [2, 5, 10]
        }
    }
}
```

```python
score = []

for model_name, mp in model_params.items():
    clf = GridSearchCV(mp['model'], mp['params'], cv = 2)
    clf.fit(x_train, y_train)
    score.append({'model' : model_name, 'score' : clf.best_score_, 'best_params' : clf.best_params_})
    
    
```
